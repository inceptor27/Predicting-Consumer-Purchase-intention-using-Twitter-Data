```python
# Predicting Consumer Purchase Intention - end-to-end script
# Run in Google Colab (recommended) or locally (Python 3.8+)
# Expects a CSV file `tweets.csv` with columns: 'tweet' and 'label' (0 or 1)
# If you want to fetch tweets with Twitter API, see the placeholder section below.

# -----------------------------
# 1) Install required packages
# -----------------------------
!pip install -q transformers sentence-transformers scikit-learn pandas matplotlib seaborn nltk vaderSentiment torch torchvision torchaudio

# -----------------------------
# 2) Imports & downloads
# -----------------------------
import os
import re
import string
import numpy as np
import pandas as pd
from tqdm.auto import tqdm

# NLP / preprocessing
import nltk
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab') # Added to resolve LookupError
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# sentiment
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

# sklearn
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, roc_curve, auc
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.neural_network import MLPClassifier
import joblib

# sentence-transformers for compact contextual embeddings
from sentence_transformers import SentenceTransformer

# plotting
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="whitegrid")
tqdm.pandas()

# -----------------------------
# 3) (Optional) Twitter collection placeholder
# -----------------------------
# If you want to collect tweets programmatically, you will need to use Twitter API v2 (bearer token) or a web-scraper.
# Below is a placeholder (do NOT run). Replace with your keys if you intend to collect live data.
"""
# Placeholder: example using tweepy (install & set up credentials)
!pip install tweepy
import tweepy
BEARER_TOKEN = "YOUR_BEARER_TOKEN"
client = tweepy.Client(bearer_token=BEARER_TOKEN)
query = "buy OR purchase OR planning to buy OR want to buy -is:retweet lang:en"
tweets = client.search_recent_tweets(query=query, tweet_fields=['created_at','public_metrics'], max_results=100)
# parse tweets into DataFrame...
"""

# -----------------------------
# 4) Load dataset
# -----------------------------
# Put your dataset in the Colab Files section or upload to the working directory.
DATA_PATH = "tweets3.csv"   # change if different
if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(f"{DATA_PATH} not found. Upload your tweets.csv to the notebook working directory.")

df = pd.read_csv(DATA_PATH)

# Rename 'text' column to 'tweet' if it exists, to match expected column name
if 'text' in df.columns and 'tweet' not in df.columns:
    df.rename(columns={'text': 'tweet'}, inplace=True)

print("Raw dataset shape:", df.shape)
display(df.head())

# Ensure the expected columns exist
if 'tweet' not in df.columns or 'label' not in df.columns:
    raise ValueError("CSV must have 'tweet' and 'label' columns. Rename your columns accordingly.")

# -----------------------------
# 5) Preprocessing functions
# -----------------------------
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
sentiment_analyzer = SentimentIntensityAnalyzer()

url_re = re.compile(r'https?://\S+|www\.\S+')
mention_re = re.compile(r'@\w+')
hashtag_re = re.compile(r'#\w+')
emoji_pattern = re.compile("["
                       u"\U0001F600-\U0001F64F"  # emoticons
                       u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                       u"\U0001F680-\U0001F6FF"  # transport & map symbols
                       u"\U0001F1E0-\U0001F1FF"  # flags
                       "]+", flags=re.UNICODE)

def clean_tweet(text):
    if pd.isna(text):
        return ""
    text = str(text)
    text = text.lower()
    text = url_re.sub('', text)
    text = mention_re.sub('', text)
    # keep hashtags text (remove # but keep word) - often product tags
    text = text.replace('#', ' ')
    text = emoji_pattern.sub('', text)
    # remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))
    # remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()
    return text

def tokenize_lemmatize(text):
    tokens = nltk.word_tokenize(text)
    tokens = [t for t in tokens if t.isalpha() and t not in stop_words]
    lemmas = [lemmatizer.lemmatize(t) for t in tokens]
    return " ".join(lemmas)

def full_preprocess(text):
    cleaned = clean_tweet(text)
    processed = tokenize_lemmatize(cleaned)
    return processed

# Apply preprocessing
df['text_clean'] = df['tweet'].progress_apply(full_preprocess)

# add sentiment score using VADER
def get_sentiment_scores(text):
    if not text or pd.isna(text):
        return 0.0
    vs = sentiment_analyzer.polarity_scores(text)
    # use compound score
    return vs['compound']

df['sentiment'] = df['tweet'].progress_apply(get_sentiment_scores)

print("After preprocessing sample:")
display(df[['tweet','text_clean','sentiment','label']].head())

# -----------------------------
# 6) Feature extraction
# -----------------------------
# TF-IDF
tfidf = TfidfVectorizer(max_features=5000, ngram_range=(1,2))
X_tfidf = tfidf.fit_transform(df['text_clean']).toarray()
print("TF-IDF shape:", X_tfidf.shape)

# Sentence-Transformer embeddings (compact contextual embeddings)
# Using a small, fast model: 'all-MiniLM-L6-v2' (good for tweets)
embed_model = SentenceTransformer('all-MiniLM-L6-v2')
embeddings = embed_model.encode(df['tweet'].tolist(), show_progress_bar=True)
print("Embeddings shape:", embeddings.shape)

# Combine features: concat TF-IDF + embeddings + sentiment (as extra column)
X = np.hstack([X_tfidf, embeddings, df['sentiment'].values.reshape(-1,1)])
y = df['label'].values
print("Combined feature shape:", X.shape)

# -----------------------------
# 7) Train / Test split
# -----------------------------
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
print("Train/Test sizes:", X_train.shape, X_test.shape)

# For some models (e.g., SVM, MLP) scale features
scaler = StandardScaler(with_mean=False)  # X_tfidf is sparse-ish; embeddings added -> avoid with_mean=True
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# -----------------------------
# 8) Models to train
# -----------------------------
models = {
    "LogisticRegression": LogisticRegression(max_iter=1000),
    "NaiveBayes(Multinomial)": MultinomialNB(),
    "SVM(RBF)": SVC(probability=True),
    "DecisionTree": DecisionTreeClassifier(max_depth=10, random_state=42),
    "MLP": MLPClassifier(hidden_layer_sizes=(128,64), max_iter=200, random_state=42)
}

# Note: MultinomialNB expects non-negative features. TF-IDF + embeddings combined might have negatives in embeddings.
# So for MultinomialNB we will re-fit using TF-IDF only.
X_tfidf_train = X_train[:, :X_tfidf.shape[1]]
X_tfidf_test  = X_test[:, :X_tfidf.shape[1]]

results = {}

for name, model in models.items():
    print(f"\nTraining {name} ...")
    if name == "NaiveBayes(Multinomial)":
        model.fit(X_tfidf_train, y_train)
        preds = model.predict(X_tfidf_test)
        probas = model.predict_proba(X_tfidf_test)[:,1]
    elif name == "SVM(RBF)":
        model.fit(X_train_scaled, y_train)
        preds = model.predict(X_test_scaled)
        probas = model.predict_proba(X_test_scaled)[:,1]
    elif name == "MLP":
        model.fit(X_train_scaled, y_train)
        preds = model.predict(X_test_scaled)
        probas = model.predict_proba(X_test_scaled)[:,1]
    else:
        # LogisticRegression, DecisionTree
        model.fit(X_train, y_train)
        preds = model.predict(X_test)
        if hasattr(model, "predict_proba"):
            probas = model.predict_proba(X_test)[:,1]
        else:
            # fallback using decision_function
            probas = model.decision_function(X_test)
            # normalize
            probas = (probas - probas.min()) / (probas.max() - probas.min() + 1e-9)
    # Evaluate
    acc = accuracy_score(y_test, preds)
    # Use average='weighted' for multiclass targets
    prec = precision_score(y_test, preds, average='weighted', zero_division=0)
    rec = recall_score(y_test, preds, average='weighted', zero_division=0)
    f1 = f1_score(y_test, preds, average='weighted', zero_division=0)
    try:
        # ROC AUC is typically for binary classification or needs different handling for multiclass
        # For simplicity here, we'll skip ROC AUC for now if target is multiclass
        if len(np.unique(y_test)) == 2:
             roc = roc_auc_score(y_test, probas)
        else:
             roc = float('nan') # Set to NaN if not binary
    except Exception:
        roc = float('nan')
    print(f"{name} -> Acc:{acc:.4f}, Prec:{prec:.4f}, Rec:{rec:.4f}, F1:{f1:.4f}, ROC_AUC:{roc:.4f}")
    results[name] = {"model": model, "accuracy": acc, "precision": prec, "recall": rec, "f1": f1, "roc_auc": roc, "preds": preds, "probas": probas}

# -----------------------------
# 9) Save the best model (example: best by F1)
# -----------------------------
best_name = max(results.keys(), key=lambda k: results[k]['f1'])
print("\nBest model by F1:", best_name, results[best_name]['f1'])
best_model = results[best_name]['model']
joblib.dump({
    "model_name": best_name,
    "model": best_model,
    "tfidf": tfidf,
    "embed_model_name": 'all-MiniLM-L6-v2',
    "scaler": scaler
}, "best_purchase_intent_model.joblib")
print("Saved best model to best_purchase_intent_model.joblib")

# -----------------------------
# 10) Confusion matrix & ROC curve (for best model)
# -----------------------------
best_preds = results[best_name]['preds']
best_probas = results[best_name]['probas']

plt.figure(figsize=(6,5))
# Pass labels to confusion_matrix to ensure correct order
cm = confusion_matrix(y_test, best_preds, labels=np.unique(y_test))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=np.unique(y_test), yticklabels=np.unique(y_test))
plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.title(f'Confusion Matrix - {best_name}')
plt.show()

# ROC Curve is primarily for binary classification.
# Skipping ROC curve plotting for multiclass case for simplicity.
if len(np.unique(y_test)) == 2:
    fpr, tpr, _ = roc_curve(y_test, best_probas)
    roc_auc = auc(fpr, tpr)
    plt.figure(figsize=(6,5))
    plt.plot(fpr, tpr, label=f'AUC = {roc_auc:.3f}')
    plt.plot([0,1],[0,1],'--')
    plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate'); plt.title(f'ROC - {best_name}')
    plt.legend(loc='lower right')
    plt.show()
else:
    print("\nSkipping ROC plot as target is multiclass.")


# -----------------------------
# 11) Example predict function for new tweets
# -----------------------------
# Load saved model (example)
bundle = joblib.load("best_purchase_intent_model.joblib")
loaded_model = bundle['model']
loaded_tfidf = bundle['tfidf']
loaded_scaler = bundle['scaler']
embed = SentenceTransformer(bundle['embed_model_name'])

def predict_single_tweet(text):
    original = text
    pre = full_preprocess(text)
    tfidf_vec = loaded_tfidf.transform([pre]).toarray()
    emb = embed.encode([original])
    sentiment = np.array([[get_sentiment_scores(original)]])
    combined = np.hstack([tfidf_vec, emb, sentiment]) # Corrected from X_tfidf, embeddings, df['sentiment']
    # If model was trained on scaled data (e.g., SVM), scale accordingly
    try:
        combined_scaled = loaded_scaler.transform(combined)
    except Exception:
        combined_scaled = combined
    # Try direct predict; if model expects different input, handle appropriately
    try:
        # Predict class directly for multiclass
        pred = loaded_model.predict(combined if combined.shape[1]==X.shape[1] else combined_scaled)[0]
        # Predict probabilities if available (useful for confidence)
        if hasattr(loaded_model, 'predict_proba'):
             # Get probability of the predicted class
             prob_index = np.where(loaded_model.classes_ == pred)[0][0]
             prob = loaded_model.predict_proba(combined if combined.shape[1]==X.shape[1] else combined_scaled)[:, prob_index][0]
        else:
             prob = None # No probability available

    except Exception:
        # fallback: logistic/regression trained on X_train (unscaled)
        pred = loaded_model.predict(combined)[0]
        prob = None
    return {"text": original, "prediction": pred, "probability": prob}

# Test with sample
examples = [
    "I'm going to buy the new iPhone next week!",
    "This product is overpriced, won't buy.",
    "I'm not sure if I want to buy this yet." # Example with potentially 'undefined' intent
]
for ex in examples:
    print(predict_single_tweet(ex))


# -----------------------------
# 12) (Optional) Simple BERT fine-tuning blueprint
# -----------------------------
# If you want to fine-tune a transformer model (e.g., 'bert-base-uncased') for classification,
# use the Hugging Face `transformers` Trainer API. This is a more involved process and requires GPU for speed.
# Here's a minimal blueprint (uncomment to use):

"""
from transformers import AutoTokenizer, AutoModelForSequenceClassification, Trainer, TrainingArguments
import datasets

tokenizer = AutoTokenizer.from_pretrained('distilbert-base-uncased')
model = AutoModelForSequenceClassification.from_pretrained('distilbert-base-uncased', num_labels=len(np.unique(df['label']))) # Adjust num_labels

# Create HF dataset
hf_ds = datasets.Dataset.from_pandas(df[['tweet','label']].rename(columns={'label':'labels'})) # Rename label to labels for HF datasets
def tokenize_fn(batch):
    return tokenizer(batch['tweet'], padding='max_length', truncation=True, max_length=128)
hf_ds = hf_ds.map(tokenize_fn, batched=True)
hf_ds = hf_ds.train_test_split(test_size=0.2, stratify_by_column='labels') # Use labels column

training_args = TrainingArguments(
    output_dir='./bert_out',
    num_train_epochs=2,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,
    evaluation_strategy='epoch',
    logging_steps=50,
    save_total_limit=1,
    load_best_model_at_end=True,
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=hf_ds['train'],
    eval_dataset=hf_ds['test'],
    tokenizer=tokenizer,
)

trainer.train()
trainer.evaluate()
"""

# -----------------------------
# 13) Final notes & tips
# -----------------------------
print("\nDone. Notes:")
print("- If you have class-imbalance, consider oversampling (SMOTE) or class_weight in models.")
print("- For production / realtime predictions, wrap `predict_single_tweet` into an API (Flask / FastAPI) and load the saved joblib file.")
print("- Replace or augment embeddings with a fine-tuned transformer for best accuracy when you have GPUs.")
print("- Keep Twitter API keys secure (do not hardcode them). Use environment variables or secrets manager.")
print("- The ROC AUC score and plot are typically for binary classification. They are skipped for multiclass targets in this script.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# -----------------------------
# 1️⃣ Load dataset
# -----------------------------
df = pd.read_csv("tweets3.csv")

# -----------------------------
# 2️⃣ Filter for iPhone only
# -----------------------------
product_name = "iPhone"
df_iphone = df[df['product'].str.lower() == product_name.lower()]

# -----------------------------
# 3️⃣ Filter for India and USA only
# -----------------------------
regions = ["India", "USA"]
df_iphone_regions = df_iphone[df_iphone['region'].isin(regions)]

# -----------------------------
# 4️⃣ Group by region and label (intent)
# -----------------------------
purchase_counts = (
    df_iphone_regions
    .groupby(['region', 'label'])
    .size()
    .reset_index(name='count')
)

purchase_counts['label'] = purchase_counts['label'].map({
    0: 'No Purchase Intent',
    1: 'Purchase Intent'
})

# -----------------------------
# 5️⃣ Plot two pie charts (side by side)
# -----------------------------
plt.figure(figsize=(10, 5))
plt.suptitle(f"Purchase Intent Comparison for {product_name}", fontsize=15, weight='bold')

for i, region in enumerate(regions):
    plt.subplot(1, 2, i + 1)
    region_data = purchase_counts[purchase_counts['region'] == region]
    plt.pie(region_data['count'],
            labels=region_data['label'],
            autopct='%1.1f%%',
            startangle=90,
            colors=sns.color_palette('Set2', 2))
    plt.title(region, fontsize=13)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

```python
# ======= BERT Fine-tuning for Purchase Intent (Colab-ready) =======
# Run in one cell. Make sure tweets.csv is uploaded first.
!pip install -q transformers datasets torch scikit-learn matplotlib seaborn tqdm

import os
import pandas as pd
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from transformers import BertTokenizer, BertForSequenceClassification, get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, roc_curve, accuracy_score
import torch.nn.functional as F
from tqdm.auto import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()

# -----------------------------
# Config
# -----------------------------
DATA_PATH = "tweets3.csv"   # change if needed
MODEL_OUT_DIR = "bert_purchase_intent_model"
PRETRAINED = "bert-base-uncased"
MAX_LEN = 128
BATCH_SIZE = 16
NUM_EPOCHS = 3
LR = 2e-5
RANDOM_SEED = 42
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", DEVICE)

# -----------------------------
# 1) Load data
# -----------------------------
if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(f"{DATA_PATH} not found. Upload your dataset.")

df = pd.read_csv(DATA_PATH)

# Rename text column to tweet
if 'text' in df.columns:
    df = df.rename(columns={'text': 'tweet'})

# Basic sanity
if 'tweet' not in df.columns or 'label' not in df.columns:
    raise ValueError("Dataset must contain 'tweet' (or text) and 'label' columns.")

# Keep only required columns
df = df[['tweet','label']].dropna().reset_index(drop=True)

# Ensure label is integer
df['label'] = df['label'].astype(int)

print("Loaded rows:", len(df))
print(df.head())

# -----------------------------
# 2) Train/test split
# -----------------------------
train_texts, val_texts, train_labels, val_labels = train_test_split(
    df['tweet'].tolist(), df['label'].tolist(),
    test_size=0.2, stratify=df['label'].tolist(), random_state=RANDOM_SEED
)

# -----------------------------
# 3) Tokenizer & Dataset
# -----------------------------
tokenizer = BertTokenizer.from_pretrained(PRETRAINED)

class TweetDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len=128):
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len
    def __len__(self):
        return len(self.texts)
    def __getitem__(self, idx):
        text = str(self.texts[idx])
        encoding = self.tokenizer(
    text,
    add_special_tokens=True,
    truncation=True,
    max_length=self.max_len,
    padding='max_length',
    return_attention_mask=True,
    return_tensors='pt'
)
        return {
            'input_ids': encoding['input_ids'].squeeze(0),
            'attention_mask': encoding['attention_mask'].squeeze(0),
            'labels': torch.tensor(self.labels[idx], dtype=torch.long)
        }

train_dataset = TweetDataset(train_texts, train_labels, tokenizer, max_len=MAX_LEN)
val_dataset   = TweetDataset(val_texts, val_labels, tokenizer, max_len=MAX_LEN)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader   = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)

# -----------------------------
# 4) Model, Optimizer, Scheduler
# -----------------------------
model = BertForSequenceClassification.from_pretrained(PRETRAINED, num_labels=2)
model.to(DEVICE)

optimizer = AdamW(model.parameters(), lr=LR, weight_decay=0.01)
total_steps = len(train_loader) * NUM_EPOCHS
scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_steps),
    num_training_steps=total_steps
)

# -----------------------------
# 5) Training & Evaluation helpers
# -----------------------------
def train_one_epoch(model, loader, optimizer, scheduler, device):
    model.train()
    losses = []
    correct = 0
    total = 0
    pbar = tqdm(loader, desc="Train", leave=False)
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        labels = batch['labels'].to(device)

        outputs = model(input_ids=input_ids, attention_mask=attention_mask, labels=labels)
        loss = outputs.loss
        logits = outputs.logits

        loss.backward()
        optimizer.step()
        scheduler.step()
        optimizer.zero_grad()

        losses.append(loss.item())
        preds = torch.argmax(logits, dim=1)
        correct += (preds == labels).sum().item()
        total += labels.size(0)

        # Show running loss and accuracy (%) in progress bar
        current_acc = (correct / total) * 100 if total > 0 else 0.0
        pbar.set_postfix({'loss': np.mean(losses), 'acc(%)': f"{current_acc:.2f}"})
    epoch_loss = np.mean(losses)
    epoch_acc = correct / total if total > 0 else 0.0
    return epoch_loss, epoch_acc

def eval_model(model, loader, device):
    model.eval()
    losses = []
    preds_all = []
    labels_all = []
    probs_all = []
    with torch.no_grad():
        pbar = tqdm(loader, desc="Eval", leave=False)
        for batch in pbar:
            input_ids = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            labels = batch['labels'].to(device)

            outputs = model(input_ids=input_ids, attention_mask=attention_mask, labels=labels)
            loss = outputs.loss
            logits = outputs.logits

            probs = F.softmax(logits, dim=1)[:, 1].cpu().numpy()  # probability of class 1
            preds = torch.argmax(logits, dim=1).cpu().numpy()

            losses.append(loss.item())
            preds_all.extend(preds.tolist())
            labels_all.extend(labels.cpu().numpy().tolist())
            probs_all.extend(probs.tolist())
    avg_loss = np.mean(losses) if losses else 0.0
    acc = accuracy_score(labels_all, preds_all) if labels_all else 0.0
    return avg_loss, acc, preds_all, labels_all, probs_all

# -----------------------------
# 6) Train Loop
# -----------------------------
train_losses, val_losses = [], []
train_accs, val_accs = [], []

for epoch in range(NUM_EPOCHS):
    print(f"\nEpoch {epoch+1}/{NUM_EPOCHS}")
    tr_loss, tr_acc = train_one_epoch(model, train_loader, optimizer, scheduler, DEVICE)
    val_loss, val_acc, val_preds, v_labels, val_probs = eval_model(model, val_loader, DEVICE)

    train_losses.append(tr_loss); val_losses.append(val_loss)
    train_accs.append(tr_acc);   val_accs.append(val_acc)

    print(f"Train loss: {tr_loss:.4f} | Train acc: {tr_acc*100:.2f}%")
    print(f"Val   loss: {val_loss:.4f} | Val   acc: {val_acc*100:.2f}%")

# -----------------------------
# 7) Final evaluation & metrics
# -----------------------------
print("\nFinal Evaluation on Validation Set:")
val_loss, val_acc, val_preds, val_labels, val_probs = eval_model(model, val_loader, DEVICE)
print(f"Validation Loss: {val_loss:.4f}")
print(f"Validation Accuracy: {val_acc:.4f} ({val_acc*100:.2f}%)")

print("\nClassification Report:\n", classification_report(val_labels, val_preds, digits=4))

# ROC AUC (binary)
try:
    roc_auc = roc_auc_score(val_labels, val_probs)
    print(f"ROC AUC: {roc_auc:.4f}")
except Exception as e:
    print("ROC AUC error:", e)
    roc_auc = None

# Confusion matrix
cm = confusion_matrix(val_labels, val_preds)
plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['No Intent','Intent'],
            yticklabels=['No Intent','Intent'])
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix (Validation)")
plt.show()

# -----------------------------
# 8) Training curves
# -----------------------------
epochs = range(1, NUM_EPOCHS+1)

plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.plot(epochs, train_losses, label='train_loss')
plt.plot(epochs, val_losses, label='val_loss')
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.legend(); plt.title("Loss Curve")

plt.subplot(1,2,2)
plt.plot(epochs, [a*100 for a in train_accs], label='train_acc(%)')
plt.plot(epochs, [a*100 for a in val_accs], label='val_acc(%)')
plt.xlabel("Epoch"); plt.ylabel("Accuracy (%)"); plt.legend(); plt.title("Accuracy Curve")
plt.tight_layout()
plt.show()

# -----------------------------
# 9) Save model & tokenizer
# -----------------------------
os.makedirs(MODEL_OUT_DIR, exist_ok=True)
model.save_pretrained(MODEL_OUT_DIR)
tokenizer.save_pretrained(MODEL_OUT_DIR)
print(f"Saved fine-tuned model & tokenizer to {MODEL_OUT_DIR}")

# -----------------------------
# 10) Utility: predict single tweet
# -----------------------------
def predict_single_tweet(text):
    model.eval()
    enc = tokenizer(
        text,
        add_special_tokens=True,
        max_length=MAX_LEN,
        truncation=True,
        padding='max_length',
        return_tensors='pt'
    )
    input_ids = enc['input_ids'].to(DEVICE)
    attention_mask = enc['attention_mask'].to(DEVICE)
    with torch.no_grad():
        out = model(input_ids=input_ids, attention_mask=attention_mask)
        logits = out.logits
        probs = F.softmax(logits, dim=1).cpu().numpy()[0]
        pred = int(np.argmax(probs))
        prob_pos = float(probs[1])
    return {
        "text": text,
        "prediction": pred,
        "probability_positive": prob_pos,
        "prediction_readable": "Purchase Intent" if pred == 1 else "No Purchase Intent"
    }

# Example predictions
print("\nExamples:")
examples = [
    "I'm planning to buy the new iPhone next month!",
    "Not going to buy that phone, too expensive."
]
for ex in examples:
    print(predict_single_tweet(ex))
```
